## Task 1 – Temperature Class

The Temperature class represents temperatures in Celsius (C), Fahrenheit (F), and Kelvin (K).

The class supports temperature conversions, comparisons across different units, and validation of input values.

The implementation includes:
- Case-insensitive temperature units.
- Validation of invalid units and temperatures below absolute zero.
- Read-only attributes for value and unit.
- Conversion methods between Celsius, Fahrenheit, and Kelvin.
- All six comparison operators, using a tolerance of 1e-9 for floating-point comparisons.
- String representations using `__str__` and `__repr__`.

The implementation is stored in `temperature.py`, and the tests are stored in `test_temperature.py`.


```python

from temperature import Temperature

# Create temperatures in different units
a = Temperature(0, "C")
b = Temperature(32, "F")
c = Temperature(273.15, "K")

# Demonstrate temperature conversions
print("Celsius to Fahrenheit:", a.to_fahrenheit())
print("Celsius to Kelvin:", a.to_kelvin())
print("Fahrenheit to Celsius:", b.to_celsius())

# Compare temperatures across different units
print("0 C == 32 F:", a == b)
print("0 C == 273.15 K:", a == c)

# Convert to a new Temperature object
converted = a.convert_to("F")
print("Converted temperature:", converted)
```


### Testing the Temperature Class

We used pytest to test the Temperature class.

The tests cover:
- Initialization, case-insensitive units, and read-only attributes.
- Invalid units and temperatures below absolute zero.
- Temperature conversions using five reference points.
- Round-trip conversions between Celsius, Fahrenheit, and Kelvin.
- Comparisons across different temperature units.
- Floating-point comparison tolerance of 1e-9.
- String representations using `__str__` and `__repr__`.

The tests are stored in `test_temperature.py`.


```python

import sys
import subprocess

# Run the Temperature tests
result = subprocess.run(
    [sys.executable, "-m", "pytest", "test_temperature.py",
     "-v", "--color=no", "--no-header"],
    capture_output=True,
    text=True
)

print(result.stdout)

if result.stderr:
    print(result.stderr)
```


### Conclusion – Task 1

The Temperature class was implemented to support Celsius, Fahrenheit, and Kelvin.

The class includes temperature conversions, input validation, read-only attributes, and all six comparison operators. A tolerance of 1e-9 is used when comparing floating-point values.

All 20 pytest tests passed, including tests for temperature conversions, invalid inputs, and comparisons across different units.

The results show that the implemented functionality works as expected for the tested cases.


## Task 2 - Fraction Class
In this task, we implemented a Python class called `Fraction` to represent fractions using a numerator and a denominator.

The class checks that both values are integers and that the denominator is greater than zero.

We implemented all six comparison operators and four additional special methods for string representation, addition, and multiplication.

The implementation is stored in `fraction.py`, and the tests are stored in `test_fraction.py`.

```python
from fraction import Fraction

# Create two fractions
a = Fraction(1, 2)
b = Fraction(1, 4)

# Demonstrate the Fraction methods
print("Fraction a:", a)
print("Fraction b:", b)
print("Addition:", a + b)
print("Multiplication:", a * b)
print("a > b:", a > b)
print("a == b:", a == b)
```


### Testing the Fraction Class

- String representations using `__str__` and `__repr__`.

The tests are stored in `test_fraction.py`.


```python

import sys
import subprocess

# Run pytest without color codes
result = subprocess.run(
    [sys.executable, "-m", "pytest", "test_fraction.py", "-v", "--color=no"],
    capture_output=True,
    text=True
)

print(result.stdout)
if result.stderr:
    print(result.stderr)
```
